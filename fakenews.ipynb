```python
import pandas as pd
import numpy as np
```

```python
from google.colab import files
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

```python
df = pd.read_csv('/content/fake_news_dataset.csv')
```

```python
print("Feature names:", df.columns[:-1].tolist())
```

```python
print("Target names:", ["Fake", "Real"])
```

```python
print("Number of samples:", df.shape[0])
```

```python
print("Number of columns:", df.shape[1])
```

```python
display(df.head())
```

```python
print(df.shape)
```

```python
print(df["label"].value_counts())
```

```python
df["title"] = df["title"].fillna("")
```

```python
df["text"] = df["text"].fillna("")
```

```python
X = df["title"] + " " + df["text"]
```

```python
y = df["label"].str.lower().map({
"fake": 0,
"real": 1
})
```

```python
print("\nShape of X:", X.shape)
```

```python
print("Shape of y:", y.shape)
```

```python
# Drop rows where y is NaN in both X and y
valid_indices = y.notna()
X_clean = X[valid_indices]
y_clean = y[valid_indices]

X_train, X_test, y_train, y_test = train_test_split(
    X_clean,
    y_clean,
    test_size=0.2,
    random_state=0,
    stratify=y_clean
)
```

```python
print("\nTraining Samples:", len(X_train))
```

```python
print("Testing Samples:", len(X_test))
```

```python
model = Pipeline([
("tfidf", TfidfVectorizer(
stop_words="english",
max_features=10000
)),
("classifier", LogisticRegression(
max_iter=1000
))
])
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
actual = np.where(
y_test == 0,
"Fake",
"Real"
)
```

```python
predicted = np.where(
y_pred == 0,
"Fake",
"Real"
)
```

```python
results = pd.DataFrame({
"Actual": actual,
"Predicted": predicted
})
```

```python
display(results.head(10))
```

```python
accuracy = accuracy_score(y_test, y_pred)
```

```python
print("\nTest Accuracy: {:.2f}%".format(
accuracy * 100
))
```

```python
predicted_label = model.predict(new_news)
```

```python
predicted_news = (
"Real" if predicted_label[0] == 1
else "Fake"
)
```
